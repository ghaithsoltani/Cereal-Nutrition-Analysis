```python
# --------------------------------------------------------------
# 4. INDEXING
# --------------------------------------------------------------
# Use the cereal 'name' column as the index since every value is unique.
cereal.set_index(['name'],drop=True, inplace=True)
```

```python
print("\n" + "=" * 70)
print("STEP 2: SELECTING COLUMNS")
print("=" * 70)
```

```python
sugars_series = cereal['grams of sugars']
nutrition_df = cereal[['calories per serving',
                       'grams of sugars'
                       ]]
print(f"# returns a Series //{sugars_series},\n# returns a DataFrame {nutrition_df}")
```

```python
# --------------------------------------------------------------
# 6. FILTERING (BOOLEAN MASKS)
# --------------------------------------------------------------
```

```python
print("\n" + "=" * 70)
print("STEP 3: FILTERING DATA")
print("=" * 70)
```

```python
kelloggs_filter = cereal['Manufacturer'] == 'Kelloggs'
low_sugar_filter = cereal['grams of sugars'] <= 5
b_filter = cereal.index.str.startswith('C')  # cereal names starting with 'C'

print("\n-- Kelloggs cereals with < 5g sugar --")
print(cereal[kelloggs_filter & low_sugar_filter][['Manufacturer', 'grams of sugars']])
print("\n-- Cereal names starting with 'C' --")
print(cereal[b_filter].index.tolist())
```

```python
# --------------------------------------------------------------
# 7. FINDING MIN / MAX ROWS
# --------------------------------------------------------------
print("\n-- Cereal with the HIGHEST calories --")
print(cereal.loc[cereal['calories per serving'] == cereal['calories per serving'].max(), :])
```

```python
print("\n-- Cereal with the LOWEST sodium --")
print(cereal.loc[cereal['milligrams of sodium'] == cereal['milligrams of sodium'].min(), :])
```

```python
manu_cal = cereal.groupby('Manufacturer')['calories per serving'].mean().sort_values()
print("\n-- Average calories per serving, by Manufacturer --")
print(manu_cal)
```

```python
manu_count = cereal.groupby('Manufacturer').size().sort_values()
print("\n-- Number of cereals per Manufacturer --")
print(manu_count)
```

```python
# --------------------------------------------------------------
# 9. CORRELATION ANALYSIS
# --------------------------------------------------------------
print("\n" + "=" * 70)
print("STEP 5: CORRELATION ANALYSIS")
print("=" * 70)

corr_value = cereal['calories per serving'].corr(cereal['grams of sugars'])
print(f"\nCorrelation (calories vs sugars): {corr_value:.3f}")

corr_matrix = cereal.corr(numeric_only=True)
print("\n-- Full correlation matrix --")
print(corr_matrix.round(2))
```

```python
# --------------------------------------------------------------
# 10. VISUALIZATION — UNIVARIATE (one column at a time)
# --------------------------------------------------------------
plt.figure(figsize=(9, 5))
plt.barh(manu_count.index, manu_count.values, color='steelblue')
plt.xlabel('Number of Cereals')
plt.ylabel('Manufacturer')
plt.title('Univariate: Number of Cereals by Manufacturer')
plt.tight_layout()
plt.savefig('/home/soltani/PycharmProjects/JupyterProject/Cereal Nutrition Analysis/images/01_univariate_bar.png', dpi=120)
plt.close()
```

```python
# --------------------------------------------------------------
# 11. VISUALIZATION — SORTED HORIZONTAL BAR CHART
# --------------------------------------------------------------
plt.figure(figsize=(9, 5))
plt.barh(manu_cal.index, manu_cal.values, color='darkorange')
plt.xlabel('Average Calories per Serving')
plt.ylabel('Manufacturer')
plt.title('Average Calories by Manufacturer (Sorted)')
plt.tight_layout()
plt.savefig('/home/soltani/PycharmProjects/JupyterProject/Cereal Nutrition Analysis/images/02_sorted_bar.png', dpi=120)
plt.close()
```
